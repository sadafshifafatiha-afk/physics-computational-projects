#1. Virtual Obstacle-Avoiding Robot

```python
import random
```

```python
import time
```

```python
class Robot:
  def move(self):
    print ("Moving forward")

  def stop(self):
    print("Stopped")

  def turn(self):
    print("Turning ")

def get_distance():
    return random.randint(1, 10)

r= Robot()

while True:
  distance = get_distance()
  print(f'Distance: {distance}')
  if distance < 2:
    r.stop()
    time.sleep(1)
  elif distance < 6:
    r.turn()
    time.sleep(1)
  else:
    r.move()
    time.sleep(1)
```

#Robot Path Planning using A*

```python
''' Grid = environment
0 = free space
1 = obstacle
Algorithm finds best path '''

import heapq
import numpy as np
import matplotlib.pyplot as plt

# Grid (0 = free, 1 = obstacle)
grid = np.array([
    [0,0,0,0,1],
    [0,1,1,0,1],
    [0,0,0,0,0],
    [1,1,0,1,0],
    [0,0,0,0,0]
])

start = (0,0)
goal = (4,4)

# Heuristic (Manhattan distance)
def h(a, b):
    return abs(a[0]-b[0]) + abs(a[1]-b[1])

# A* Algorithm
def astar(grid, start, goal):
    rows, cols = grid.shape
    open_set = []
    heapq.heappush(open_set, (0, start))

    came_from = {}
    g_score = {start: 0}

    while open_set:
        _, current = heapq.heappop(open_set)

        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            return path[::-1]

        neighbors = [(0,1),(1,0),(0,-1),(-1,0)]
        for dx, dy in neighbors:
            neighbor = (current[0]+dx, current[1]+dy)

            if 0 <= neighbor[0] < rows and 0 <= neighbor[1] < cols:
                if grid[neighbor] == 1:
                    continue

                temp_g = g_score[current] + 1

                if neighbor not in g_score or temp_g < g_score[neighbor]:
                    g_score[neighbor] = temp_g
                    f = temp_g + h(neighbor, goal)
                    heapq.heappush(open_set, (f, neighbor))
                    came_from[neighbor] = current

    return None

path = astar(grid, start, goal)

# Plot
plt.imshow(grid, cmap='gray')

if path:
    y, x = zip(*path)
    plt.plot(x, y, color='red', linewidth=2)

plt.scatter(start[1], start[0], color='green', label='Start')
plt.scatter(goal[1], goal[0], color='blue', label='Goal')

plt.legend()
plt.title("A* Path Planning")
plt.show()
```
