#DC Motor Speed Control Simulation (with PID)

```python
import numpy as np
import matplotlib.pyplot as plt

# Time
t = np.linspace(0, 10, 100)

# Target speed
setpoint = 50

# Initial speed
speed = 0

# PID constants
Kp = 0.8
Ki = 0.2
Kd = 0.05

error_sum = 0
prev_error = 0

speeds = []

for time in t:
    error = setpoint - speed # Calculate error between target and current speed
    error_sum += error #add past errors (Integral)
    d_error = error - prev_error #change in error (Derivative)

    # PID formula
    control = Kp*error + Ki*error_sum + Kd*d_error

    speed += control * 0.1 # 0.1 cause real motors act slow

    speeds.append(speed)
    prev_error = error

plt.plot(t, speeds, label="Motor Speed")
plt.axhline(setpoint, color='r', linestyle='--', label="Target")
plt.xlabel("Time")
plt.ylabel("Speed")
plt.title("PID Motor Control Simulation")
plt.legend()
plt.show()
```
